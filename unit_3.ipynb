### 3.1 For Loops

```python
import math

my_list = [10, 5, 3]
total = 1

for i in my_list:
  total *= i

print(total)
print(math.prod(my_list))
```

### 3.2 Python Range() Function

```python
stocks_name = ['Apple', 'Samsung', 'Alibaba', 'Novo Nordisk', 'Naspers']
stock_prices = [142, 45, 12, 125, 30]

total_stocks = len(stocks_name)

for i in reversed(range(total_stocks)):
  print(f'Index = {i}')
  print(f'The price of {stocks_name[i]} stock = ${stock_prices[i]}')
```

```python
for i in reversed(range(0, total_stocks ,2)):
  print(f'Index = {i}')
  print(f'The price of {stocks_name[i]} stock = ${stock_prices[i]}')
```

### 3.3 While Loops

```python
pv = float(input('Enter the amount of funds you would like to invest today in dollars (Present Value (PV)'))
r = float(input('Enter the annual interest rate'))
fv = float(input('Enter the total dollar amount you would like to have in the future (Future Value (FV))'))
year = 0

print(f'The Present Value (PV) in year {year} (now) = ${pv}')
while pv < fv:
  year += 1
  pv = pv + (r*pv)
  print(f'FV in Year {year} = ${round(pv,2)}')

print(f'It would take around {year} years to reach a fitire value greater than ${fv}')
```

### 3.4 Functions

```python
def EAR(i, m):
    return ((1 + (i/m))**m) - 1

quarterly_ear = round(EAR(0.1, 4), 4)
monthly_ear = round(EAR(0.1, 12), 4)

print(1000*quarterly_ear)
print(1000*monthly_ear)
```
