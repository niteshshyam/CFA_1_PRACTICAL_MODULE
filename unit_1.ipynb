### 1.3 Get User Input

```python
# CAPM: E(Ri) = Bi*(E(Rm)-Rf) + Rf
stock = input("Enter a stock")
B_i = float(input(f"Enter a Beta value for a {stock}"))
R_f = float(input("Enter the risk-free rate (%)"))
E_Rm = float(input("Etner the broad market return (%)"))

E_Ri = B_i*(E_Rm - R_f) + R_f
print(f"The Expected Return [E(Ri)] on {stock} = {round(E_Ri, 2)}%")
```
