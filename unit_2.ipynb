### 2.5 Conditional Statements

```python
# DDM: P = D0(1+g)/(r-g)
price_per_share = 64
minimum_rate_of_return_r = 0.14
dividend_per_share_D0 = 2
growth_rate_g = 0.1

intrinsic_value_P = (dividend_per_share_D0*(1+growth_rate_g))/(minimum_rate_of_return_r-growth_rate_g)
print(intrinsic_value_P) 
```

```python
if intrinsic_value_P < price_per_share:
    print("Overvalued")
elif intrinsic_value_P > price_per_share:
    print("Undervalued")
else:
    print("Fairly Valued")
```

```python
dividend_per_share_D0 = float(input("Enter this year's dividend: "))
growth_rate_g = float(input("Enter the company's expected dividend growth: "))
minimum_rate_of_return_r = float(input("Enter the company's cost of equity: "))
price_per_share = float(input("Enter the current stock price: "))

intrinsic_value_P = (dividend_per_share_D0*(1+growth_rate_g))/(minimum_rate_of_return_r-growth_rate_g)
print(intrinsic_value_P) 

if intrinsic_value_P < price_per_share:
    print("Overvalued, do not invest in this stock")
elif intrinsic_value_P > price_per_share:
    print("Undervalued")
else:
    print("Fairly Valued")
```
