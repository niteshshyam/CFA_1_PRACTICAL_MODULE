### 4.4 DataFrames and Functions

```python
investors_df = pd.read_csv('investors_data.csv')
investors_df
```

```python
def rankings(portfolio_size):
    if portfolio_size < 100000:
        return 'Low'
    elif portfolio_size >= 100000 and portfolio_size <= 500000:
        return 'Medium'
    else:
        return 'High'
```

```python
investors_df['Rank'] = investors_df['Portfolio Size'].apply(rankings)
investors_df
```

### 4.5 DataFrames Concatenation and Merging
